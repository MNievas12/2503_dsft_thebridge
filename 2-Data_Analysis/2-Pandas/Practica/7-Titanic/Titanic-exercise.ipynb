21.  Group `titanic` by `'pclass'` and aggregate by the columns `age` & `fare`, by `max` and `median` and assign it to `by_class`


```python
def q75(x):
    return x.quantile(0.75)
```

```python
by_class = titanic.groupby('pclass')[['age','fare','survived']].aggregate(['max','median','mean','min', q75])
by_class
```

```python
by_class.columns
```
